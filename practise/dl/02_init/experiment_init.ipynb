```python
from init_experiment import activation_stds

DEPTH, WIDTH = 30, 256
show = [1, 5, 10, 20, 30]  # после каких слоёв печатать std

results = {}
for scheme in ("xavier", "he"):
    for act in ("tanh", "relu"):
        results[(scheme, act)] = activation_stds(DEPTH, WIDTH, scheme, act)

print(f"{'init + act':<16}" + "".join(f"{'layer ' + str(k):>12}" for k in show))
for (scheme, act), stds in results.items():
    row = "".join(f"{stds[k - 1]:>12.2e}" for k in show)
    print(f"{scheme + ' + ' + act:<16}{row}")
```

```python
import matplotlib.pyplot as plt

# цвет = схема инициализации, тип линии = активация
colors = {"xavier": "#2a78d6", "he": "#eb6834"}
styles = {"relu": "-", "tanh": "--"}
layers = range(1, DEPTH + 1)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.axhline(1.0, color="#b0aea5", lw=1, zorder=0)  # std входа
for (scheme, act), stds in results.items():
    label = f"{scheme} + {act}"
    ax.plot(layers, stds, color=colors[scheme], ls=styles[act], lw=2, label=label)
    ax.annotate(label, (DEPTH, stds[-1]), xytext=(6, 0), textcoords="offset points",
                va="center", fontsize=9, color="#3d3d3a")

ax.set_yscale("log")
ax.set_xlabel("номер слоя")
ax.set_ylabel("std активаций (log)")
ax.set_title(f"std активаций по слоям, width = {WIDTH}")
ax.set_xlim(1, DEPTH + 6)
ax.set_xticks([1, 5, 10, 15, 20, 25, 30])
ax.grid(True, which="major", color="#e8e6dc", lw=0.8)
for side in ("top", "right"):
    ax.spines[side].set_visible(False)
ax.legend(frameon=False, loc="lower left")
plt.tight_layout()
plt.show()
```
