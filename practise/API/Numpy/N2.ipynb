```python
import numpy as np
```

```python
x = np.array([[2, 10], 
              [7, 20], 
              [9, 30]
              ])

# строки, где значение в первом столбце больше 5
mask = x[:, 0] > 5
rows = x[mask]
print(rows, rows.shape)

# В полученной выборке замени значения второго столбца на -1
rows[:, 1] = -1
print(rows, rows.shape)

# x – не изменится
print(x)
```
