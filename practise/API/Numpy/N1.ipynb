```python
import numpy as np
```

```python
x = np.array([
    [[1, 2, 3]],
    [[1, 2, 3]]
])

print(x.shape) # 2 блока 1 список внутри блока 3 числа
print(x.ndim) # 3 оси, так как три столбца
print(x.size) # 6 числе
```

Число осей равно длине кортежа формы. Если добавить по числу в каждый внутренний список, форма станет (2, 1, 4), а осей по-прежнему будет три.

```python
a = np.array([[10, 20, 30, 40]])
b = np.array([a])

print(a.shape, a.ndim, a.size)
print(b.shape, b.ndim, b.size)
```

```python
x = np.array([2.9, -2.9, 0.5], dtype=np.float32)
y = x.astype(np.int32)

print(x.dtype, y.dtype)
print(y.tolist())
```

```python
x = np.array([10, 20])
a = np.asarray(x)
b = np.array(x)

x[0] = 99

print(a.tolist())
print(b.tolist())
```

```python
x = np.array([
    [1, 2, 3],
    [2, 2, 36],
    [3, 1, 33],
    [4, 25, 3],
    [5, 2, 31]
], dtype=np.float32)

print(x.shape, x.dtype) #новая память выделиться, потому что мы создали другую матрицу с другой размерность и типом данных
```

В примере выше NumPy выделяет память при создании массива из списков и сразу записывает значения в формате float32.
