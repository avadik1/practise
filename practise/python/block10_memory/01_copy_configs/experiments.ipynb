# copy vs deepcopy — сетка конфигов

Перед запуском ячейки с `make_configs_shallow` **сначала запиши предсказание**.

```python
from configs import make_configs

base = {"model": {"lr": 1e-3, "optimizer": "adam"}, "epochs": 10}
cfgs = make_configs(base, [1e-2, 1e-3, 1e-4])
print([c["model"]["lr"] for c in cfgs])
print(base["model"]["lr"])
```

## Предсказание

Заменяю в `make_configs` способ копирования на поверхностный — `base.copy()`. Что напечатает `base["model"]["lr"]` и что будет в списке lr у `cfgs`? Почему?

_Мой прогноз:_

print([c["model"]["lr"] for c in cfgs]) – [1e-4, 1e-4, 1e-4]


print(base["model"]["lr"]) – 1e-4

**Что увидел:**

все правильно

```python
def make_configs_shallow(base, lrs):
    out = []
    for lr in lrs:
        cfg = base.copy()
        cfg["model"]["lr"] = lr
        out.append(cfg)
    return out

base = {"model": {"lr": 1e-3, "optimizer": "adam"}, "epochs": 10}
cfgs = make_configs_shallow(base, [1e-2, 1e-3, 1e-4])
print([c["model"]["lr"] for c in cfgs])
print(base["model"]["lr"])
```

## Вывод (одной фразой)

Почему при `.copy()` сломались и `base`, и сами конфиги?

- Потому что вложенные изменяемые объекты меняются вместе с оригиналом, а в итоговый список пойдет всегда последняя "версия"
