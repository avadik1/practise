## Прогноз (до запуска). Пусть cpu x1 = T сек
- cpu x4 ≈ 4T
- io  x1 ≈ 2сек
- io  x4 ≈ 2сек

```python
import time, threading

def cpu_task(n=30_000_000):
    s = 0
    for i in range(n):
        s += i

def io_task():
    time.sleep(2)

def run(target, k):
    ts = [threading.Thread(target=target) for _ in range(k)]
    t0 = time.perf_counter()
    for t in ts: t.start()
    for t in ts: t.join()
    return time.perf_counter() - t0

print("cpu x1:", run(cpu_task, 1))
print("cpu x4:", run(cpu_task, 4))
print("io  x1:", run(io_task, 1))
print("io  x4:", run(io_task, 4))
```

## Вывод
Прогноз совпал полностью: cpu x4 ≈ 4.0T (3.48 с при T = 0.86 с), io x1 ≈ io x4 ≈ 2.0 с. На CPU-bound потоки не ускоряют, потому что GIL даёт исполнять байткод только одному потоку за раз. На IO-bound `time.sleep` отпускает GIL, поэтому 4 ожидания идут параллельно.
